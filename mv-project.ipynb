```python
# ============================================================================
# PART 1: IMPORTS AND SETUP
# ============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import os
import shutil
from pathlib import Path

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

# Suppress warnings
import warnings
warnings.filterwarnings('ignore')

print("="*70)
print("HAND GESTURE RECOGNITION WITH XAI - TRAINING ON KAGGLE")
print("="*70)

# Check GPU
print("\n📊 Environment Check:")
print(f"TensorFlow version: {tf.__version__}")
gpus = tf.config.list_physical_devices('GPU')
print(f"GPU Available: {len(gpus) > 0}")
if gpus:
    print(f"GPU Devices: {gpus}")
print("="*70)

# ============================================================================
# PART 2: CONFIGURATION
# ============================================================================

# Paths
BASE_PATH = '/kaggle/input/leapgestrecog/leapGestRecog'
OUTPUT_PATH = 'leap_organized'

# Hyperparameters
IMG_SIZE = 128
BATCH_SIZE = 64
EPOCHS = 40
LEARNING_RATE = 0.001
TEST_SPLIT = 0.2
VALIDATION_SPLIT = 0.2

print("\n📋 Configuration:")
print(f"Image Size: {IMG_SIZE}x{IMG_SIZE}")
print(f"Batch Size: {BATCH_SIZE}")
print(f"Epochs: {EPOCHS}")
print(f"Learning Rate: {LEARNING_RATE}")
print("="*70)

# ============================================================================
# PART 3: DATASET PREPARATION
# ============================================================================

print("\n" + "="*70)
print("PREPARING LEAP GESTURE DATASET")
print("="*70)

def prepare_leap_dataset(base_path, output_path, test_split=0.2):
    """
    Reorganize Leap dataset structure:
    
    Input: leapGestRecog/00/01_palm/, 01/01_palm/, ...
    Output: leap_organized/train/palm/, test/palm/
    """
    
    print("\n📁 Scanning dataset structure...")
    
    # Create output directories
    train_path = os.path.join(output_path, 'train')
    test_path = os.path.join(output_path, 'test')
    os.makedirs(train_path, exist_ok=True)
    os.makedirs(test_path, exist_ok=True)
    
    # Get subject folders (00, 01, 02, ...)
    subject_folders = sorted([f for f in os.listdir(base_path) 
                             if os.path.isdir(os.path.join(base_path, f)) 
                             and not f.startswith('.')])
    
    print(f"✅ Found {len(subject_folders)} subject folders: {subject_folders}")
    
    # Get gesture classes from first subject
    first_subject = os.path.join(base_path, subject_folders[0])
    gesture_folders = sorted([f for f in os.listdir(first_subject) 
                             if os.path.isdir(os.path.join(first_subject, f))
                             and not f.startswith('.')])
    
    # Extract clean gesture names
    gesture_names = []
    for g in gesture_folders:
        if '_' in g:
            name = g.split('_', 1)[1]
        else:
            name = g
        gesture_names.append(name)
    
    gesture_names = sorted(list(set(gesture_names)))
    
    print(f"✅ Found {len(gesture_names)} gesture classes:")
    print(f"   {gesture_names}")
    
    # Collect all images per gesture
    gesture_images = {name: [] for name in gesture_names}
    
    print("\n📸 Collecting images from all subjects...")
    for subject in subject_folders:
        subject_path = os.path.join(base_path, subject)
        
        for gesture_folder in os.listdir(subject_path):
            gesture_path = os.path.join(subject_path, gesture_folder)
            
            if not os.path.isdir(gesture_path):
                continue
            
            # Extract gesture name
            if '_' in gesture_folder:
                gesture_name = gesture_folder.split('_', 1)[1]
            else:
                gesture_name = gesture_folder
            
            if gesture_name not in gesture_names:
                continue
            
            # Get images
            images = [os.path.join(gesture_path, f) 
                     for f in os.listdir(gesture_path)
                     if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
            
            gesture_images[gesture_name].extend(images)
    
    # Split and copy files
    print("\n📊 Splitting into train/test sets...")
    
    total_train = 0
    total_test = 0
    
    for gesture_name, images in gesture_images.items():
        n_images = len(images)
        
        if n_images == 0:
            print(f"   ⚠️  {gesture_name}: No images found, skipping")
            continue
        
        # Create directories
        train_gesture = os.path.join(train_path, gesture_name)
        test_gesture = os.path.join(test_path, gesture_name)
        os.makedirs(train_gesture, exist_ok=True)
        os.makedirs(test_gesture, exist_ok=True)
        
        # Split
        train_imgs, test_imgs = train_test_split(
            images, test_size=test_split, random_state=42
        )
        
        # Copy train images
        for i, img_path in enumerate(train_imgs):
            ext = os.path.splitext(img_path)[1]
            dest = os.path.join(train_gesture, f"{gesture_name}_train_{i}{ext}")
            shutil.copy2(img_path, dest)
        
        # Copy test images
        for i, img_path in enumerate(test_imgs):
            ext = os.path.splitext(img_path)[1]
            dest = os.path.join(test_gesture, f"{gesture_name}_test_{i}{ext}")
            shutil.copy2(img_path, dest)
        
        total_train += len(train_imgs)
        total_test += len(test_imgs)
        
        print(f"   ✅ {gesture_name.ljust(15)}: {len(train_imgs):>4} train, {len(test_imgs):>4} test")
    
    print(f"\n✅ Total: {total_train} train, {total_test} test images")
    print("="*70)
    
    return output_path, gesture_names

# Prepare dataset
ORGANIZED_PATH, class_names = prepare_leap_dataset(BASE_PATH, OUTPUT_PATH, TEST_SPLIT)
TRAIN_PATH = os.path.join(ORGANIZED_PATH, 'train')
TEST_PATH = os.path.join(ORGANIZED_PATH, 'test')
num_classes = len(class_names)

print(f"\n✅ Dataset ready!")
print(f"✅ Classes ({num_classes}): {class_names}")
print(f"✅ Train path: {TRAIN_PATH}")
print(f"✅ Test path: {TEST_PATH}")

# ============================================================================
# PART 4: DATA PREPROCESSING AND AUGMENTATION
# ============================================================================

print("\n" + "="*70)
print("CREATING DATA GENERATORS WITH PREPROCESSING")
print("="*70)

def preprocess_image(img):
    """
    Preprocess with CLAHE enhancement
    """
    # Ensure RGB
    if len(img.shape) == 2:
        img = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
    
    # Convert to uint8
    if img.max() <= 1.0:
        img_uint8 = (img * 255).astype(np.uint8)
    else:
        img_uint8 = img.astype(np.uint8)
    
    # CLAHE enhancement
    lab = cv2.cvtColor(img_uint8, cv2.COLOR_RGB2LAB)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
    l = clahe.apply(l)
    enhanced = cv2.merge([l, a, b])
    enhanced = cv2.cvtColor(enhanced, cv2.COLOR_LAB2RGB)
    
    return enhanced.astype(np.float32) / 255.0

# Training data generator with augmentation
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_image,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.15,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest',
    validation_split=VALIDATION_SPLIT
)

# Load generators
train_generator = train_datagen.flow_from_directory(
    TRAIN_PATH,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    shuffle=True
)

val_generator = train_datagen.flow_from_directory(
    TRAIN_PATH,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    shuffle=False
)

print(f"✅ Train samples: {train_generator.samples}")
print(f"✅ Validation samples: {val_generator.samples}")
print(f"✅ Steps per epoch: {train_generator.samples // BATCH_SIZE}")

# ============================================================================
# PART 5: VISUALIZE SAMPLE IMAGES
# ============================================================================

print("\n📸 Creating sample visualization...")

fig, axes = plt.subplots(3, 5, figsize=(15, 9))
axes = axes.ravel()

images, labels = next(train_generator)

for i in range(15):
    if i < len(images):
        axes[i].imshow(images[i])
        class_idx = np.argmax(labels[i])
        axes[i].set_title(class_names[class_idx], fontsize=11, fontweight='bold')
        axes[i].axis('off')

plt.suptitle('Sample Training Images (After Preprocessing)', 
            fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('sample_images.png', dpi=150, bbox_inches='tight')
print("✅ Saved: sample_images.png")
plt.show()

# ============================================================================
# PART 6: BUILD CNN MODEL
# ============================================================================

print("\n" + "="*70)
print("BUILDING CNN MODEL")
print("="*70)

model = keras.Sequential([
    # Input
    layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)),
    
    # Block 1
    layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),
    
    # Block 2
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),
    
    # Block 3
    layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),
    
    # Block 4 (for Grad-CAM)
    layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='last_conv'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),
    
    # Dense layers
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(256, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(num_classes, activation='softmax', name='output')
])

# Compile
optimizer = keras.optimizers.Adam(learning_rate=LEARNING_RATE)
model.compile(
    optimizer=optimizer,
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("\n📊 Model Summary:")
model.summary()
print(f"\n✅ Total parameters: {model.count_params():,}")

# ============================================================================
# PART 7: SETUP CALLBACKS
# ============================================================================

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=8,
        restore_best_weights=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=4,
        min_lr=1e-7,
        verbose=1
    ),
    keras.callbacks.ModelCheckpoint(
        'best_gesture_model.h5',
        monitor='val_accuracy',
        save_best_only=True,
        verbose=1
    )
]

# ============================================================================
# PART 8: TRAIN MODEL
# ============================================================================

print("\n" + "="*70)
print("STARTING TRAINING")
print("="*70)
print(f"⏱️  Estimated time: 30-40 minutes with GPU")
print(f"📊 Epochs: {EPOCHS}")
print(f"📦 Batch size: {BATCH_SIZE}")
print(f"🎯 Learning rate: {LEARNING_RATE}")
print("="*70 + "\n")

history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=val_generator,
    callbacks=callbacks,
    verbose=1
)

print("\n✅ Training completed!")

# ============================================================================
# PART 9: PLOT TRAINING HISTORY
# ============================================================================

print("\n📊 Creating training history plots...")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy
ax1.plot(history.history['accuracy'], label='Train', linewidth=2.5, color='#2E86AB')
ax1.plot(history.history['val_accuracy'], label='Validation', linewidth=2.5, color='#A23B72')
ax1.set_title('Model Accuracy', fontsize=14, fontweight='bold')
ax1.set_xlabel('Epoch', fontsize=12)
ax1.set_ylabel('Accuracy', fontsize=12)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)

# Loss
ax2.plot(history.history['loss'], label='Train', linewidth=2.5, color='#2E86AB')
ax2.plot(history.history['val_loss'], label='Validation', linewidth=2.5, color='#A23B72')
ax2.set_title('Model Loss', fontsize=14, fontweight='bold')
ax2.set_xlabel('Epoch', fontsize=12)
ax2.set_ylabel('Loss', fontsize=12)
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('training_history.png', dpi=150, bbox_inches='tight')
print("✅ Saved: training_history.png")
plt.show()

# ============================================================================
# PART 10: EVALUATE MODEL
# ============================================================================

print("\n" + "="*70)
print("EVALUATING MODEL")
print("="*70)

# Evaluate
val_loss, val_accuracy = model.evaluate(val_generator, verbose=0)
print(f"\n📊 Validation Results:")
print(f"   Loss: {val_loss:.4f}")
print(f"   Accuracy: {val_accuracy*100:.2f}%")

# Get predictions
print("\n🔍 Generating predictions...")
val_generator.reset()
predictions = model.predict(val_generator, verbose=1)
y_pred = np.argmax(predictions, axis=1)
y_true = val_generator.classes

# Classification report
print("\n📋 Classification Report:")
print(classification_report(y_true, y_pred, target_names=class_names, digits=3))

# ============================================================================
# PART 11: CONFUSION MATRIX
# ============================================================================

print("\n📊 Creating confusion matrix...")

cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(12, 10))
sns.heatmap(cm, annot=True, fmt='d', cmap='YlOrRd',
           xticklabels=class_names,
           yticklabels=class_names,
           cbar_kws={'label': 'Count'},
           linewidths=0.5,
           linecolor='gray')
plt.title('Confusion Matrix', fontsize=16, fontweight='bold', pad=20)
plt.ylabel('True Label', fontsize=13)
plt.xlabel('Predicted Label', fontsize=13)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight')
print("✅ Saved: confusion_matrix.png")
plt.show()

# ============================================================================
# PART 12: CONFIDENCE DISTRIBUTION
# ============================================================================

print("\n📊 Analyzing confidence distribution...")

max_confidences = np.max(predictions, axis=1)

plt.figure(figsize=(10, 6))
plt.hist(max_confidences, bins=50, edgecolor='black', alpha=0.7, color='#2E86AB')
plt.axvline(x=0.85, color='#06A77D', linestyle='--', linewidth=2.5, 
           label='High Confidence (>85%)')
plt.axvline(x=0.70, color='#F18F01', linestyle='--', linewidth=2.5, 
           label='Medium Confidence (>70%)')
plt.title('Prediction Confidence Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Maximum Confidence Score', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig('confidence_distribution.png', dpi=150, bbox_inches='tight')
print("✅ Saved: confidence_distribution.png")
plt.show()

# ============================================================================
# PART 13: CORRECT PREDICTIONS VISUALIZATION
# ============================================================================

print("\n📸 Creating correct predictions visualization...")

correct_mask = y_pred == y_true
correct_indices = np.where(correct_mask)[0]
confidences = np.max(predictions, axis=1)

high_conf_correct = correct_indices[confidences[correct_indices] > 0.85]

if len(high_conf_correct) >= 20:
    sample_indices = np.random.choice(high_conf_correct, 20, replace=False)
else:
    sample_indices = high_conf_correct

fig, axes = plt.subplots(4, 5, figsize=(20, 16))
fig.suptitle('✅ CORRECT PREDICTIONS WITH HIGH CONFIDENCE', 
            fontsize=18, fontweight='bold', y=0.995, color='#06A77D')

for idx, ax in enumerate(axes.flat):
    if idx < len(sample_indices):
        sample_idx = sample_indices[idx]
        
        # Get image
        batch_idx = sample_idx // val_generator.batch_size
        in_batch_idx = sample_idx % val_generator.batch_size
        val_generator.reset()
        for _ in range(batch_idx + 1):
            images, labels = next(val_generator)
        img = images[in_batch_idx]
        
        pred_class = y_pred[sample_idx]
        true_class = y_true[sample_idx]
        conf = confidences[sample_idx]
        
        ax.imshow(img)
        ax.set_title(f'Gesture: {class_names[true_class]}\n'
                    f'Predicted: {class_names[pred_class]}\n'
                    f'Confidence: {conf*100:.1f}%',
                    fontsize=10, color='#06A77D', fontweight='bold')
        ax.axis('off')
    else:
        ax.axis('off')

plt.tight_layout()
plt.savefig('validation_correct_predictions.png', dpi=150, bbox_inches='tight')
print("✅ Saved: validation_correct_predictions.png")
plt.show()

# ============================================================================
# PART 14: INCORRECT PREDICTIONS VISUALIZATION
# ============================================================================

print("\n📸 Creating incorrect predictions visualization...")

incorrect_indices = np.where(~correct_mask)[0]

if len(incorrect_indices) > 0:
    sample_incorrect = np.random.choice(
        incorrect_indices, 
        min(20, len(incorrect_indices)), 
        replace=False
    )
    
    fig, axes = plt.subplots(4, 5, figsize=(20, 16))
    fig.suptitle('❌ INCORRECT PREDICTIONS - ERROR ANALYSIS', 
                fontsize=18, fontweight='bold', y=0.995, color='#D62828')
    
    for idx, ax in enumerate(axes.flat):
        if idx < len(sample_incorrect):
            sample_idx = sample_incorrect[idx]
            
            batch_idx = sample_idx // val_generator.batch_size
            in_batch_idx = sample_idx % val_generator.batch_size
            val_generator.reset()
            for _ in range(batch_idx + 1):
                images, labels = next(val_generator)
            img = images[in_batch_idx]
            
            pred_class = y_pred[sample_idx]
            true_class = y_true[sample_idx]
            conf = confidences[sample_idx]
            
            ax.imshow(img)
            ax.set_title(f'True: {class_names[true_class]}\n'
                        f'Predicted: {class_names[pred_class]}\n'
                        f'Confidence: {conf*100:.1f}%',
                        fontsize=10, color='#D62828', fontweight='bold')
            ax.axis('off')
        else:
            ax.axis('off')
    
    plt.tight_layout()
    plt.savefig('validation_incorrect_predictions.png', dpi=150, bbox_inches='tight')
    print("✅ Saved: validation_incorrect_predictions.png")
    plt.show()

# ============================================================================
# PART 15: PER-CLASS ACCURACY
# ============================================================================

print("\n📊 Creating per-class accuracy analysis...")

class_accuracies = []
class_samples = []

for i in range(num_classes):
    mask = y_true == i
    if np.sum(mask) > 0:
        acc = np.mean(y_pred[mask] == y_true[mask]) * 100
        class_accuracies.append(acc)
        class_samples.append(np.sum(mask))
    else:
        class_accuracies.append(0)
        class_samples.append(0)

sorted_indices = np.argsort(class_accuracies)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Best classes
best_count = min(len(class_names), 10)
best_indices = sorted_indices[-best_count:]
bars1 = ax1.barh([class_names[i] for i in best_indices],
                 [class_accuracies[i] for i in best_indices],
                 color='#06A77D', alpha=0.8, edgecolor='black', linewidth=1.5)
ax1.set_xlabel('Accuracy (%)', fontsize=12, fontweight='bold')
ax1.set_title(f'Top {best_count} Best Performing Gestures', fontsize=14, fontweight='bold')
ax1.set_xlim(0, 100)
ax1.grid(axis='x', alpha=0.3)

for i, idx in enumerate(best_indices):
    ax1.text(class_accuracies[idx] + 1, i, 
            f'n={class_samples[idx]}', va='center', fontsize=9)

# Worst classes
worst_count = min(len(class_names), 10)
worst_indices = sorted_indices[:worst_count]
bars2 = ax2.barh([class_names[i] for i in worst_indices],
                 [class_accuracies[i] for i in worst_indices],
                 color='#D62828', alpha=0.8, edgecolor='black', linewidth=1.5)
ax2.set_xlabel('Accuracy (%)', fontsize=12, fontweight='bold')
ax2.set_title(f'Top {worst_count} Challenging Gestures', fontsize=14, fontweight='bold')
ax2.set_xlim(0, 100)
ax2.grid(axis='x', alpha=0.3)

for i, idx in enumerate(worst_indices):
    ax2.text(class_accuracies[idx] + 1, i, 
            f'n={class_samples[idx]}', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('validation_per_class_accuracy.png', dpi=150, bbox_inches='tight')
print("✅ Saved: validation_per_class_accuracy.png")
plt.show()

# ============================================================================
# PART 16: CONFIDENCE VS ACCURACY ANALYSIS
# ============================================================================

print("\n📊 Creating confidence vs accuracy analysis...")

conf_bins = [0, 0.5, 0.7, 0.85, 1.0]
conf_labels = ['<50%', '50-70%', '70-85%', '>85%']
bin_accuracies = []
bin_counts = []

for i in range(len(conf_bins) - 1):
    mask = (max_confidences >= conf_bins[i]) & (max_confidences < conf_bins[i+1])
    if np.sum(mask) > 0:
        acc = np.mean(correct_mask[mask]) * 100
        bin_accuracies.append(acc)
        bin_counts.append(np.sum(mask))
    else:
        bin_accuracies.append(0)
        bin_counts.append(0)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

colors = ['#D62828', '#F18F01', '#F9C74F', '#06A77D']

# Accuracy by confidence
bars = ax1.bar(conf_labels, bin_accuracies, color=colors, alpha=0.8, 
              edgecolor='black', linewidth=1.5)
ax1.set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
ax1.set_xlabel('Confidence Range', fontsize=12, fontweight='bold')
ax1.set_title('Accuracy vs Confidence Level', fontsize=14, fontweight='bold')
ax1.set_ylim(0, 105)
ax1.grid(axis='y', alpha=0.3)

for bar, count in zip(bars, bin_counts):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 2,
            f'n={count}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# Sample distribution
bars2 = ax2.bar(conf_labels, bin_counts, color=colors, alpha=0.8, 
               edgecolor='black', linewidth=1.5)
ax2.set_ylabel('Number of Samples', fontsize=12, fontweight='bold')
ax2.set_xlabel('Confidence Range', fontsize=12, fontweight='bold')
ax2.set_title('Sample Distribution by Confidence', fontsize=14, fontweight='bold')
ax2.grid(axis='y', alpha=0.3)

total = sum(bin_counts)
for bar, count in zip(bars2, bin_counts):
    height = bar.get_height()
    pct = (count / total * 100) if total > 0 else 0
    ax2.text(bar.get_x() + bar.get_width()/2., height + max(bin_counts)*0.02,
            f'{pct:.1f}%', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('validation_confidence_analysis.png', dpi=150, bbox_inches='tight')
print("✅ Saved: validation_confidence_analysis.png")
plt.show()

# ============================================================================
# PART 17: SAVE MODEL AND CLASS NAMES
# ============================================================================

print("\n" + "="*70)
print("SAVING MODEL AND FILES")
print("="*70)

# Save model
model.save('gesture_model_final.h5')
print("✅ Saved: gesture_model_final.h5")

# Save class names
import json
class_mapping = {i: name for i, name in enumerate(class_names)}
with open('class_names.json', 'w') as f:
    json.dump(class_mapping, f, indent=2)
print("✅ Saved: class_names.json")

# ============================================================================
# PART 18: FINAL SUMMARY
# ============================================================================

print("\n" + "="*70)
print("VALIDATION SUMMARY")
print("="*70)

accuracy = np.mean(correct_mask) * 100

print(f"\n✅ Overall Accuracy: {accuracy:.2f}%")

print(f"\n📊 Confidence Distribution:")
for label, count in zip(conf_labels, bin_counts):
    pct = (count / total * 100) if total > 0 else 0
    print(f"   {label.ljust(10)}: {count:>5} samples ({pct:>5.1f}%)")

print(f"\n🎯 High Confidence Predictions:")
high_conf_mask = max_confidences > 0.85
high_conf_acc = np.mean(correct_mask[high_conf_mask]) * 100 if np.sum(high_conf_mask) > 0 else 0
print(f"   Samples with >85% confidence: {np.sum(high_conf_mask)} ({np.sum(high_conf_mask)/len(max_confidences)*100:.1f}%)")
print(f"   Accuracy on high confidence: {high_conf_acc:.2f}%")

print(f"\n⚠️  Low Confidence Predictions:")
low_conf_mask = max_confidences < 0.70
low_conf_acc = np.mean(correct_mask[low_conf_mask]) * 100 if np.sum(low_conf_mask) > 0 else 0
print(f"   Samples with <70% confidence: {np.sum(low_conf_mask)} ({np.sum(low_conf_mask)/len(max_confidences)*100:.1f}%)")
if np.sum(low_conf_mask) > 0:
    print(f"   Accuracy on low confidence: {low_conf_acc:.2f}%")

print(f"\n🏆 Best Performing Classes (Top 5):")
for i, idx in enumerate(sorted_indices[-5:][::-1]):
    print(f"   {i+1}. {class_names[idx].ljust(15)}: {class_accuracies[idx]:.1f}% (n={class_samples[idx]})")

print(f"\n⚠️  Most Challenging Classes (Top 5):")
for i, idx in enumerate(sorted_indices[:5]):
    if class_samples[idx] > 0:
        print(f"   {i+1}. {class_names[idx].ljust(15)}: {class_accuracies[idx]:.1f}% (n={class_samples[idx]})")

print("\n" + "="*70)
print("📥 FILES TO DOWNLOAD FROM KAGGLE OUTPUT")
print("="*70)
print("\n✅ ESSENTIAL (for local inference):")
print("   1. best_gesture_model.h5")
print("   2. class_names.json")
print("\n📊 FOR REPORT/PRESENTATION:")
print("   3. sample_images.png")
print("   4. training_history.png")
print("   5. confusion_matrix.png")
print("   6. confidence_distribution.png")
print("   7. validation_correct_predictions.png")
print("   8. validation_incorrect_predictions.png")
print("   9. validation_per_class_accuracy.png")
print("   10. validation_confidence_analysis.png")
print("\n💡 HOW TO DOWNLOAD:")
print("   - Click 'Output' tab on the right")
print("   - Click the download icon (⬇) next to each file")
print("="*70)

print("\n" + "="*70)
print("🎉 TRAINING AND VALIDATION COMPLETE!")
print("="*70)
print(f"\n✅ Final Validation Accuracy: {val_accuracy*100:.2f}%")
print(f"✅ Model: Leap Gesture Recognition")
print(f"✅ Classes: {num_classes} gestures")
print(f"✅ Total training time: ~{len(history.history['accuracy'])} epochs")
print(f"✅ Ready for local deployment!")
print("\n" + "="*70)
print("NEXT STEPS:")
print("="*70)
print("1. Download 'best_gesture_model.h5' and 'class_names.json'")
print("2. Place them in your local project folder")
print("3. Run the local inference script: python local_inference.py")
print("4. Use the PNG images in your presentation/report")
print("="*70)
```
